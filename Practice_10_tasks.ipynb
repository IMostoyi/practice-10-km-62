### **Завдання 1. Ланцюжок вузлів (1 бал)**

##### **Частина 1. Нерекурсивний конструктор**

Реалізуйте нерекурсивну функцію `make_link(value, tail=None)`, яка повертає кортеж `(value, tail)`. Значення `tail` — це інший вузол такого самого вигляду або `None`, що позначає кінець ланцюжка. Якщо `tail` не задано, створюється останній вузол.

Наприклад, `make_link(3, make_link(2, make_link(1)))` має повернути `(3, (2, (1, None)))`.

Функція має лише створювати новий вузол, зберігаючи посилання на переданий хвіст. Обходити ланцюжок не потрібно.

```
# ВАШ КОД ТУТ

# ПЕРЕВІРКА

assert make_link(5) == (5, None)
assert make_link('кінець', None) == ('кінець', None)
chain = make_link(3, make_link(2, make_link(1)))
assert chain == (3, (2, (1, None)))
assert chain[1][1][0] == 1

tail = make_link(10)
head = make_link(20, tail)
assert head[1] is tail
assert tail == (10, None)
print('Усі перевірки частини 1 пройдено!')
```

##### **Частина 2. Рекурсивне сумування ланцюжка**

Реалізуйте рекурсивну функцію `sum_links(chain)`, яка приймає **ланцюжок, створений функцією `make_link` із частини 1**, і повертає суму значень усіх його вузлів. Для цієї частини вважайте, що значення вузлів — цілі числа.

Порожній ланцюжок позначається `None`, його сума дорівнює `0`. Для непорожнього ланцюжка додайте значення поточного вузла до суми його хвоста, обчисленої рекурсивно. Цикли та готову функцію `sum` використовувати не можна.

Приклад: `sum_links(make_link(3, make_link(2, make_link(1))))` має повернути `6`. Вхідну структуру змінювати не можна.

```
# ВАШ КОД ТУТ

# ПЕРЕВІРКА

assert sum_links(None) == 0
assert sum_links(make_link(7)) == 7
assert sum_links(make_link(0)) == 0
chain = make_link(3, make_link(2, make_link(1)))
assert sum_links(chain) == 6
assert chain == (3, (2, (1, None))), 'Вхідний ланцюжок змінено'
assert sum_links(make_link(2, make_link(-5, make_link(3)))) == 0
print('Усі перевірки частини 2 пройдено!')
```

Порожній список має довжину `0`. Для непорожнього списку врахуйте перший елемент і рекурсивно знайдіть довжину решти. Вкладений список рахується як один елемент: для `[1, [2, 3]]` результат дорівнює `2`.

Використовувати `len` у власній реалізації не можна; у перевірках її застосовують для порівняння. Для відокремлення першого елемента від решти можна використовувати зріз `items[1:]`.

```
# ВАШ КОД ТУТ

# ПЕРЕВІРКА

for items in ([], [True], [1, 2, 3], [1, 0, 3], ['', 'текст'], [None], [[1], 'так', -2], [True, []]):
    original = items.copy()
    result = recursive_len(items)
    assert result == len(items), f'Помилка для {items!r}'
    assert type(result) is int, 'Результат має бути цілим числом'
    assert items == original, 'Початковий список змінено'
print('Усі перевірки завдання 2 пройдено!')
```

### **Завдання 3. Біноміальні коефіцієнти (1 бал)**

Реалізуйте рекурсивну функцію `binomial(n, k)`, яка обчислює кількість способів обрати $k$ елементів із $n$ за правилом трикутника Паскаля:

$$
C(n,k) = \begin{cases}
1, & k = 0 \text{ або } k = n; \\
C(n-1,k-1) + C(n-1,k), & 0 < k < n.
\end{cases}
$$

Аргументи — цілі числа, $0 \le k \le n \le 10$. Зокрема, $C(0,0)=1$. Використовувати готові функції обчислення біноміальних коефіцієнтів або формулу через факторіали не можна.

```
# ВАШ КОД ТУТ

# ПЕРЕВІРКА

cases = [
    (0, 0, 1), (1, 0, 1), (1, 1, 1),
    (5, 0, 1), (5, 5, 1), (5, 2, 10),
    (6, 3, 20), (7, 2, 21), (7, 5, 21), (10, 5, 252),
]
for n, k, expected in cases:
    assert binomial(n, k) == expected, f'Помилка для n={n}, k={k}'
print('Усі перевірки завдання 3 пройдено!')
```
